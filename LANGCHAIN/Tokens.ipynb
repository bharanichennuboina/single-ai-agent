```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("Bert-base-uncased")
```

```python
text = "Good morning my buddies.."
```

```python
tokens = tokenizer.tokenize(text)
```

```python
print("Tokens:",tokens)
```

```python
len(tokens)
```

```python
encoded = tokenizer(text)
```

```python
print("Input ID:",encoded["input_ids"])
```

```python
vocab = tokenizer.get_vocab
```

```python
type(vocab)
```

```python
len(vocab)
```

```python
import tiktoken
encodeing = tiktoken.encoding_for_model('gpt-4o')
```

```python
text = "Good morning my buddies.."
```

```python
tokens = encodeing.encode(text)
```

```python
print("Tokens:",tokens)
```

```python
print("\nIndividual tokens:")
for token in tokens:
    print(encodeing.decode([token]),end="|")
```
