```python
import os 
from langchain.chat_models import init_chat_model
os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
model = init_chat_model(
    'openai/gpt-oss-120b',
    model_provider = 'groq')
response = model.invoke("do u have a girlfriend?")
print(response.content) 
```

```python
import os 
from langchain.chat_models import init_chat_model
os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
model = init_chat_model(
    'openai/gpt-oss-120b',
    model_provider = 'groq',
    temperature = 0.2,
    max_tokens = 100,
    model_kwargs = {
        "top_p":0.1
    }
    ) 

response = model.invoke("who are u?")
print(response.content) 
```
