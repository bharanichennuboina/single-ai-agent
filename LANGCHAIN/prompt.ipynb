```python
import os

from groq import Groq

client = Groq(
    api_key=os.environ.get("GROQ_API_KEY"),
)

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "how many ex-girlfriends do u have?",
        }
    ],
    model="openai/gpt-oss-120b",
)

print(chat_completion.choices[0].message.content)
```

```python
import os

from groq import Groq
zero_shot = """Classify the movie REview"""
movie_review1 = "Tell me the Animal Movie Review."

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": zero_shot
        },
        {
            "role":"user",
            "content":movie_review1,
        }
    ],
    model="openai/gpt-oss-120b",
)

print(chat_completion.choices[0].message.content)
```
